```python
%sql
CREATE OR REPLACE VIEW hetul_catalog_ott_02.silver.masked_silver_users AS
SELECT
  CASE
    WHEN current_user() = 'admin' THEN user_id
    WHEN current_user() IN ('data_scientist','data_engineer') THEN sha2(user_id, 256)
    ELSE '***MASKED***'
  END AS user_id,
  plan_type,
  billing_cycle,
  CASE
    WHEN current_user() IN ('admin','data_scientist') THEN payment_status
    ELSE '***MASKED***'
  END AS payment_status,
  churn_flag
FROM hetul_catalog_ott_02.silver.silver_users;
```

```python
%sql
CREATE OR REPLACE VIEW hetul_catalog_ott_02.silver.masked_silver_events AS
SELECT
  CASE
    WHEN current_user() = 'admin' THEN event_id
    WHEN current_user() IN ('data_scientist','data_engineer') THEN sha2(event_id, 256)
    ELSE '***MASKED***'
  END AS event_id,
  CASE
    WHEN current_user() = 'admin' THEN user_id
    WHEN current_user() IN ('data_scientist','data_engineer') THEN sha2(user_id, 256)
    ELSE '***MASKED***'
  END AS user_id,
  content_id,
  start_time,
  end_time,
  device,
  geo,
  bitrate,
  errors,
  session_status,
  watch_span
FROM hetul_catalog_ott_02.silver.silver_events;
```
